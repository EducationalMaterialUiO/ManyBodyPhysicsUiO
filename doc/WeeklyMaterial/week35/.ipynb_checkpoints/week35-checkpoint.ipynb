```python
import numpy as np
from itertools import combinations, permutations
np.set_printoptions(precision=6, suppress=True, linewidth=120)
rng = np.random.default_rng(2026)
```

## 1. Fermion operators as matrices in a finite Fock space

With $n$ single-particle states the Fock space has $2^n$ basis states, one per occupation pattern. Orbital $p$ occupies bit $p$ of the basis-state index, and the sign $(-1)^{\#\{\text{occupied orbitals below } p\}}$ keeps the anticommutation relations exact (this is the Jordan-Wigner construction, which we will meet again later in the course). We build $a_p$ and $a_p^\dagger$ as $2^n\times 2^n$ matrices and verify **all three anticommutation relations of the lecture**, for every pair of orbitals:

$$\{a_p^\dagger,a_q^\dagger\}=\{a_p,a_q\}=0,\qquad \{a_p^\dagger,a_q\}=\delta_{pq}.$$

```python
class FockSpace:
    '''Fermion creation/annihilation operators as matrices on 2^n Fock states.'''

    def __init__(self, n_orbitals):
        self.n = n_orbitals
        self.dim = 1 << n_orbitals

    def annihilate(self, p):
        M = np.zeros((self.dim, self.dim))
        for state in range(self.dim):
            if state & (1 << p):
                sign = (-1) ** bin(state & ((1 << p) - 1)).count("1")
                M[state ^ (1 << p), state] = sign
        return M

    def create(self, p):
        return self.annihilate(p).T

    def vacuum(self):
        v = np.zeros(self.dim)
        v[0] = 1.0
        return v

    def string(self, ops):
        '''Matrix of a product of operators; ops = [(index, dagger), ...].'''
        M = np.eye(self.dim)
        for index, dagger in ops:
            M = M @ (self.create(index) if dagger else self.annihilate(index))
        return M


def anticommutator(A, B):
    return A @ B + B @ A

n = 4
space = FockSpace(n)
a = [space.annihilate(p) for p in range(n)]
adag = [space.create(p) for p in range(n)]

ok_cc = all(np.allclose(anticommutator(adag[p], adag[q]), 0)
            for p in range(n) for q in range(n))
ok_aa = all(np.allclose(anticommutator(a[p], a[q]), 0)
            for p in range(n) for q in range(n))
ok_mixed = all(np.allclose(anticommutator(adag[p], a[q]),
                           (p == q) * np.eye(space.dim))
               for p in range(n) for q in range(n))
print(f"{{a_p^+, a_q^+}} = 0 for all p,q          : {ok_cc}")
print(f"{{a_p,   a_q  }} = 0 for all p,q          : {ok_aa}")
print(f"{{a_p^+, a_q  }} = delta_pq for all p,q   : {ok_mixed}")
print(f"a_p |0> = 0 for all p                   : "
      f"{all(np.allclose(a[p] @ space.vacuum(), 0) for p in range(n))}")
```

## 2. Slater determinants as operator strings

An antisymmetric state is a string of creation operators on the vacuum,
$$|\alpha_1\alpha_2\dots\alpha_N\rangle_{\rm AS} = a_{\alpha_1}^\dagger a_{\alpha_2}^\dagger\cdots a_{\alpha_N}^\dagger|0\rangle.$$
We check, from the matrices alone, the two defining properties the Slater determinant had in week 34: interchanging two labels flips the sign (antisymmetry), and a repeated label gives the zero vector (the Pauli principle). We also check the sign rule used in pause-to-think #2: annihilating an orbital that sits $k$ places from the front costs $(-1)^k$.

```python
vac = space.vacuum()

state_012 = adag[0] @ adag[1] @ adag[2] @ vac      # |012>
state_102 = adag[1] @ adag[0] @ adag[2] @ vac      # |102>: one transposition
state_002 = adag[0] @ adag[0] @ adag[2] @ vac      # repeated orbital

print("|102> == -|012>            :", np.allclose(state_102, -state_012))
print("|002> == 0  (Pauli)        :", np.allclose(state_002, 0))
print("<012|012> = ", state_012 @ state_012, " (normalised for free)")

# pause #2: a_1 |012> = -|02>, and a_1^+ a_1 |012> = +|012>
state_02 = adag[0] @ adag[2] @ vac
print("a_1 |012> == -|02>         :", np.allclose(a[1] @ state_012, -state_02))
print("a_1^+ a_1 |012> == +|012>  :",
      np.allclose(adag[1] @ a[1] @ state_012, state_012))
```

## 3. The number operator and one-body operators

The number operator $\hat N=\sum_p a_p^\dagger a_p$ must count particles, and a one-body operator takes the form
$$\hat H_0=\sum_{pq}\langle p|\hat h_0|q\rangle\, a_p^\dagger a_q .$$
We draw a random Hermitian $h$, build $\hat H_0$ as a matrix, and check three statements from the lecture: $\hat N$ has the determinants as eigenstates with the particle number as eigenvalue; $[\hat N,\hat H_0]=0$ (pause #3: each term creates one particle and destroys one); and the expectation value in a determinant is the sum of the diagonal matrix elements over the occupied orbitals, $\langle\Phi|\hat H_0|\Phi\rangle=\sum_{\mu\,\in\,{\rm occ}} \langle\mu|\hat h_0|\mu\rangle$ -- the result week 34 needed a page of permutation arguments to establish.

```python
Nhat = sum(adag[p] @ a[p] for p in range(n))

H = rng.normal(size=(n, n))
h = (H + H.T) / 2                       # a random real symmetric h_pq
H0 = sum(h[p, q] * adag[p] @ a[q] for p in range(n) for q in range(n))

print("N |012> = 3 |012>          :", np.allclose(Nhat @ state_012, 3 * state_012))
print("N |02>  = 2 |02>           :", np.allclose(Nhat @ state_02, 2 * state_02))
print("[N, H0] = 0                :", np.allclose(Nhat @ H0 - H0 @ Nhat, 0))

expval = state_012 @ H0 @ state_012
print(f"<012|H0|012> = {expval:.6f}")
print(f"h_00+h_11+h_22 = {h[0,0]+h[1,1]+h[2,2]:.6f}   agree: "
      f"{np.isclose(expval, h[0,0]+h[1,1]+h[2,2])}")

# and a first off-diagonal Condon-Slater-type element for free:
state_013 = adag[0] @ adag[1] @ adag[3] @ vac
print(f"<013|H0|012> = {state_013 @ H0 @ state_012:.6f}  vs  h_32 = {h[3,2]:.6f}"
      f"   (states differing in one orbital: a single matrix element)")
```

## 4. The two-body operator: results of section 3.2

The lecture *stated* (derivations in the book, section 3.2) that
$$\hat H_I=\frac12\sum_{pqrs}\langle pq|\hat v|rs\rangle\, a_p^\dagger a_q^\dagger a_s a_r
=\frac14\sum_{pqrs}\langle pq|\hat v|rs\rangle_{\rm AS}\, a_p^\dagger a_q^\dagger a_s a_r ,$$
with the annihilation operators in **reversed** order. We build a random interaction with the correct symmetry $\langle pq|\hat v|rs\rangle=\langle qp|\hat v|sr\rangle$, construct both forms as matrices, and verify that

* the $\tfrac12$-form and the $\tfrac14$-AS-form are the **same matrix**;
* $\langle ij|\hat H_I|ij\rangle=\langle ij|\hat v|ij\rangle-\langle ij|\hat v|ji\rangle$, direct minus exchange;
* with the *wrong* operator order $a_r a_s$ every matrix element flips sign (pause #4).

```python
# random two-body matrix elements with <pq|v|rs> = <qp|v|sr>
v = rng.normal(size=(n, n, n, n))
v = (v + v.transpose(1, 0, 3, 2)) / 2
v_as = v - v.transpose(0, 1, 3, 2)      # <pq|v|rs>_AS = <pq|v|rs> - <pq|v|sr>

HI_half = sum(0.5 * v[p, q, r, s] * adag[p] @ adag[q] @ a[s] @ a[r]
              for p in range(n) for q in range(n)
              for r in range(n) for s in range(n))
HI_quarter = sum(0.25 * v_as[p, q, r, s] * adag[p] @ adag[q] @ a[s] @ a[r]
                 for p in range(n) for q in range(n)
                 for r in range(n) for s in range(n))
HI_wrong = sum(0.5 * v[p, q, r, s] * adag[p] @ adag[q] @ a[r] @ a[s]
               for p in range(n) for q in range(n)
               for r in range(n) for s in range(n))

print("1/2-form == 1/4-AS-form            :", np.allclose(HI_half, HI_quarter))
print("[N, H_I] = 0                       :",
      np.allclose(Nhat @ HI_half - HI_half @ Nhat, 0))

i, j = 0, 2
ket_ij = adag[i] @ adag[j] @ vac
me = ket_ij @ HI_half @ ket_ij
print(f"<{i}{j}|H_I|{i}{j}>                        = {me:+.6f}")
print(f"<{i}{j}|v|{i}{j}> - <{i}{j}|v|{j}{i}> (dir - exch) = "
      f"{v[i,j,i,j]-v[i,j,j,i]:+.6f}   agree: "
      f"{np.isclose(me, v[i,j,i,j]-v[i,j,j,i])}")
print("wrong order a_r a_s == -H_I        :", np.allclose(HI_wrong, -HI_half))
```

## 5. Normal ordering and contractions, checked as matrix identities

Normal ordering puts every creation operator to the left, with a sign $(-1)^\kappa$ from the interchanges,
$$N[a_p a_q^\dagger]=-a_q^\dagger a_p,\qquad N[a_p^\dagger a_q]=a_p^\dagger a_q .$$
The key property is $\langle 0|N[\cdots]|0\rangle=0$ for every non-empty product, and the **contraction** is the difference between a product and its normal-ordered product -- a *number*. We check the four elementary contractions of the lecture as full matrix identities: $a_pa_q^\dagger-N[a_pa_q^\dagger]=\delta_{pq}\,\mathbb{1}$, and zero for the other three pairs.

```python
def normal_order(ops):
    '''(sign, reordered): creation operators first, relative orders kept.'''
    creators = [o for o in ops if o[1]]
    annihilators = [o for o in ops if not o[1]]
    swaps = sum(1 for i, o in enumerate(ops) if not o[1]
                for p in ops[i + 1:] if p[1])
    return (-1) ** swaps, tuple(creators + annihilators)


def normal_product(space, ops):
    sign, reordered = normal_order(ops)
    return sign * space.string(reordered)


c = lambda p: (p, True)     # a_p^dagger
d = lambda p: (p, False)    # a_p

# <0| N[...] |0> = 0 for a batch of random non-empty strings
ok = True
for _ in range(200):
    length = rng.integers(1, 7)
    ops = [(int(rng.integers(0, n)), bool(rng.integers(0, 2)))
           for _ in range(length)]
    ok &= np.isclose(vac @ normal_product(space, ops) @ vac, 0)
print("<0|N[any non-empty string]|0> = 0  :", ok)

# the four elementary contractions, as matrices
I = np.eye(space.dim)
for p in range(n):
    for q in range(n):
        c1 = space.string([d(p), c(q)]) - normal_product(space, [d(p), c(q)])
        c2 = space.string([d(p), d(q)]) - normal_product(space, [d(p), d(q)])
        c3 = space.string([c(q), d(p)]) - normal_product(space, [c(q), d(p)])
        c4 = space.string([c(p), c(q)]) - normal_product(space, [c(p), c(q)])
        assert np.allclose(c1, (p == q) * I) and np.allclose(c2, 0) \
           and np.allclose(c3, 0) and np.allclose(c4, 0)
print("a_p a_q^+ - N[a_p a_q^+] = delta_pq * identity, other three pairs = 0 : True")
```

## 6. Wick's theorem: brute force against pairings

Now the theorem itself, for vacuum expectation values:
$$\langle 0|xyz\cdots w|0\rangle=\sum_{\text{fully contracted}}(-1)^{\nu}\prod(\text{contractions}),$$
with $\nu$ the number of crossings of the contraction lines. We implement the sum over perfect matchings (a compact version of `BookPrograms/chapter03/wick.py`) and compare it, for a batch of random operator strings, against the direct matrix evaluation $\langle 0|\,\text{string}\,|0\rangle$ in the Fock space. Odd strings must give zero on both routes.

```python
def perfect_matchings(indices):
    if not indices:
        yield []
        return
    first, rest = indices[0], indices[1:]
    for k in range(len(rest)):
        for tail in perfect_matchings(rest[:k] + rest[k + 1:]):
            yield [(first, rest[k])] + tail


def matching_sign(pairs):
    '''(-1)^(number of crossings of the contraction lines).'''
    crossings = sum(1 for (A, B), (C, D) in combinations(pairs, 2)
                    if A < C < B < D or C < A < D < B)
    return (-1) ** crossings


def contraction_value(x, y):
    '''<0| x y |0>: only a_p a_q^dagger with p == q survives.'''
    return 1.0 if (not x[1]) and y[1] and x[0] == y[0] else 0.0


def vev_wick(ops):
    ops = tuple(ops)
    if len(ops) % 2:
        return 0.0
    total = 0.0
    for pairs in perfect_matchings(list(range(len(ops)))):
        value = 1.0
        for A, B in pairs:
            value *= contraction_value(ops[A], ops[B])
            if value == 0.0:
                break
        total += matching_sign(pairs) * value
    return total


def vev_matrix(space, ops):
    v0 = space.vacuum()
    return v0 @ space.string(ops) @ v0


ok, tested = True, 0
for _ in range(300):
    length = int(rng.integers(2, 9))
    ops = [(int(rng.integers(0, n)), bool(rng.integers(0, 2)))
           for _ in range(length)]
    ok &= np.isclose(vev_wick(ops), vev_matrix(space, ops))
    tested += 1
print(f"Wick == matrix evaluation on {tested} random strings (M = 2..8): {ok}")

print("example <0|a_1 a_0 a_0^+ a_1^+|0> :",
      vev_wick([d(1), d(0), c(0), c(1)]),
      "  (the nested pairing, +d_00 d_11)")
print("example <0|a_0 a_1 a_0^+ a_1^+|0> :",
      vev_wick([d(0), d(1), c(0), c(1)]),
      " (the crossed pairing, -d_00 d_11)")
print("example <0|a_3 a_2 a_0^+ a_1^+|0> :",
      vev_wick([d(3), d(2), c(0), c(1)]),
      "  (all indices differ: every contraction vanishes)")
```

## 7. Counting the work: the table of the slides

For strings of $n$ annihilation operators followed by $n$ creation operators ($M=2n$, all indices distinct) we reproduce the cost table shown in the lecture: the number of pairings $(M-1)!!$ that Wick's theorem enumerates, the number of recursive calls the brute-force anticommutation route makes, and the number $n!$ of pairings that survive the elementary contractions. The recursion is the push-to-the-right recipe of the lecture, implemented literally on *symbolic* indices (all distinct, so no accidental Kronecker deltas).

```python
calls = [0]

def brute_force_calls(ops):
    '''Count the recursive calls of the anticommutation route (symbolic indices).'''
    ops = tuple(ops)
    calls[0] += 1
    if len(ops) == 0 or len(ops) % 2 == 1:
        return
    if ops[0][1] or not ops[-1][1]:      # starts with a^+, or ends with a
        return
    for k in range(len(ops) - 1):
        left, right = ops[k], ops[k + 1]
        if not left[1] and right[1]:     # a_p a_q^dagger found
            brute_force_calls(ops[:k] + ops[k + 2:])            # delta term
            brute_force_calls(ops[:k] + (right, left) + ops[k + 2:])  # swap
            return


def double_factorial(m):
    out, k = 1, m - 1
    while k > 1:
        out, k = out * k, k - 2
    return out


def survivors(m):
    '''Pairings that survive: each a must contract with an a^+ to its right.'''
    half = m // 2
    ops = [(f"x{k}", False) for k in range(half)] + \
          [(f"y{k}", True) for k in range(half)]
    count = 0
    for pairs in perfect_matchings(list(range(m))):
        if all((not ops[A][1]) and ops[B][1] for A, B in pairs):
            count += 1
    return count


print(f"{'M':>4s} {'pairings (M-1)!!':>18s} {'recursive calls':>17s} "
      f"{'survivors n!':>14s}")
for m in (2, 4, 6, 8, 10):
    half = m // 2
    ops = [(f"x{k}", False) for k in range(half)] + \
          [(f"y{k}", True) for k in range(half)]
    calls[0] = 0
    brute_force_calls(ops)
    print(f"{m:4d} {double_factorial(m):18d} {calls[0]:17d} {survivors(m):14d}")
print()
print("The theorem does not win on raw counts (see M = 10); it wins because")
print("the pairings are known in advance and the survivors can be written")
print("down directly, signs included.  See the discussion in the lecture.")
```

## 8. The overlap $\langle rs|pq\rangle$ and the determinant hiding in six operators

Finally the two showpieces of the lecture. First the four-operator overlap, evaluated over **all** index combinations and compared with the Wick result
$$\langle rs|pq\rangle=\delta_{rp}\delta_{sq}-\delta_{sp}\delta_{rq}.$$
Then the six-operator expectation value of pause #6: $\langle 0|a_{c}a_{b}a_{a}\,a_d^\dagger a_e^\dagger a_f^\dagger|0\rangle$ equals the $3\times3$ determinant $\det(\delta_{i\mu})$ built from the overlaps -- Wick's theorem generates determinants without being told about them.

```python
# <rs|pq> for every index combination, matrices vs Wick formula
ok = all(np.isclose(vev_matrix(space, [d(s), d(r), c(p), c(q)]),
                    (r == p) * (s == q) - (s == p) * (r == q))
         for p in range(n) for q in range(n)
         for r in range(n) for s in range(n))
print("<rs|pq> == d_rp d_sq - d_sp d_rq for all p,q,r,s :", ok)

# six operators: <0| a_k3 a_k2 a_k1 a_m1^+ a_m2^+ a_m3^+ |0> == det(delta)
ok6 = True
for _ in range(50):
    ks = rng.integers(0, n, size=3)
    ms = rng.integers(0, n, size=3)
    ops = [d(int(ks[2])), d(int(ks[1])), d(int(ks[0])),
           c(int(ms[0])), c(int(ms[1])), c(int(ms[2]))]
    delta = np.array([[1.0 * (ki == mi) for mi in ms] for ki in ks])
    ok6 &= np.isclose(vev_matrix(space, ops), np.linalg.det(delta))
print("six-operator VEV == det(delta_ij) on 50 random index sets :", ok6)
```

## 9. The three-level model of the exercise session

Exercise 2 of the week-35 set introduces a three-level model with doubly degenerate single-particle levels $p=1,2,3$ (spacing $d$, first level at energy $d$), two fermions, and an interaction whose matrix elements between the allowed pair states -- both particles in the same level -- all equal $-g$. With the pair configurations $|p\rangle$ as basis, the Hamiltonian matrices of parts b) and c) are

$$H^{(2)}=\begin{pmatrix}2d-g & -g\\ -g & 4d-g\end{pmatrix},\qquad
H^{(3)}=\begin{pmatrix}2d-g & -g & -g\\ -g & 4d-g & -g\\ -g & -g & 6d-g\end{pmatrix}.$$

Part c) invites a numerical diagonalisation, so here it is: eigenvalues and ground-state components for a range of interaction strengths $g/d$. The admixture of the excited configurations grows with $g/d$ -- the quantitative version of the statement that the reliability of a single-determinant ansatz depends on the ratio of interaction strength to level spacing. (Derivations and the analytic $2\times2$ solution: do them on paper first; the answers are in chapter 2 of the lecture notes.)

```python
d = 1.0

def H2(g):
    return np.array([[2*d - g, -g],
                     [-g, 4*d - g]])

def H3(g):
    return np.array([[2*d - g, -g, -g],
                     [-g, 4*d - g, -g],
                     [-g, -g, 6*d - g]])

print("three levels (part c), d = 1:")
print(f"{'g':>5s} {'E0':>9s} {'E1':>9s} {'E2':>9s}"
      f"   ground state (c1, c2, c3)")
for g in (0.0, 0.2, 0.5, 1.0, 2.0):
    E, C = np.linalg.eigh(H3(g))
    c0 = C[:, 0] * np.sign(C[0, 0])          # fix the overall phase
    print(f"{g:5.1f} {E[0]:9.4f} {E[1]:9.4f} {E[2]:9.4f}"
          f"   ({c0[0]:+.4f}, {c0[1]:+.4f}, {c0[2]:+.4f})")

print()
print("two levels (part b), same g values -- compare with your analytic result:")
for g in (0.0, 0.2, 0.5, 1.0, 2.0):
    E, C = np.linalg.eigh(H2(g))
    c0 = C[:, 0] * np.sign(C[0, 0])
    print(f"  g = {g:3.1f}: E0 = {E[0]:8.4f},  ground state = "
          f"({c0[0]:+.4f}, {c0[1]:+.4f}),  admixture of |p=2>^2 = {c0[1]**2:.4f}")
```
